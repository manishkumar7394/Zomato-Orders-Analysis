# Data Analyst

## Zomato Revenue Analysis 

```python
# 1. What is the total revenue?
revenue= df['order_amount'].sum()
print(f"Total Revenue: {revenue}")
```

```python
#2. What is the monthly revenue trend?
df['order_timestamp']= pd.to_datetime(df['order_timestamp'])
df['month'] = df['order_timestamp'].dt.to_period('M')
print(df.groupby('month')['order_amount'].sum().reset_index())
```

```python
#3. Which city contributes the highest revenue?

city_revenue= df.groupby('city_y')['order_amount'].sum().sort_values(ascending=False).reset_index()
city_revenue
```

```python
#4. Which payment mode generates the most revenue?
payment_revenue =df.groupby('payment_mode')['order_amount'].sum().sort_values(ascending=False).reset_index()
payment_revenue
```

```python
#5. What is the Average Order Value (AOV)?
Aov = df['order_amount'].mean()
print(f"Average Order value: {Aov:.2f}")
```

## Customer Analysis 

```python
#1. Who are the top 10 customers by revenue?\
customer_revenue =df.groupby(['customer_name', 'customer_id'])['order_amount'].sum().sort_values(ascending=False).head(10).reset_index()
customer_revenue
```

```python
# 2. What percentage of revenue comes from top customers?
df['customer_revenue'] = df.groupby('customer_id')['order_amount'].transform('sum')
top_customers_revenue = df[df['customer_revenue'] > df['customer_revenue'].quantile(0.9)]['order_amount'].sum()
total_revenue = df['order_amount'].sum()
percentage_top_customers = (top_customers_revenue / total_revenue) * 100
print(f"Percentage of revenue from top customers: {percentage_top_customers:.2f}%")
```

```python
# 3. Which acquisition channel brings the highest-value customers?
df['customer_value'] = df.groupby('customer_id')['order_amount'].transform('sum')
acquisition_channel_value = df.groupby('acquisition_channel')['customer_value'].sum()
highest_value_channel = acquisition_channel_value.idxmax()
print(f"Acquisition channel with highest-value customers: {highest_value_channel}")
```

```python
# 4. How many repeat customers do we have?
df['customer_order_count'] = df.groupby('customer_id')['order_id'].transform('count')
repeat_customers_count = df[df['customer_order_count'] > 1]['customer_id'].nunique()
print(f"Number of repeat customers: {repeat_customers_count}")
```

## Restaurant Performance

```python
# 1. Which restaurants generate the highest revenue?
restuarant_revenues = df.groupby("restaurant_name")['order_amount'].sum().sort_values(ascending=False).reset_index(name='total_revenue')
restuarant_revenues
```

```python
# 2. Which restaurants receive the most orders?
restaurant_orders = df.groupby("restaurant_name")['order_id'].count().sort_values(ascending=False)
restaurant_orders 
```

```python
# 3. Which cuisines are most popular?
cuisine_orders = df.groupby('cuisine')['order_id'].count().sort_values(ascending=False).reset_index()
cuisine_orders 
```

```python
# 4. Do highly-rated restaurants generate more revenue?
highly_rated_revenue = df.groupby('restaurant_name').agg({'avg_rating': 'mean', 'order_amount': 'sum' }).sort_values(by='avg_rating', ascending=False).head(3).reset_index()
highly_rated_revenue 
```

```python
# 5. Last 5 Restaurants
last_5_restaurants = df.groupby('restaurant_name').size().sort_values(ascending=False).tail(5).reset_index(name='order_count')
last_5_restaurants 
```

## Coupon Analysis

```python
# A coupon is considered used when discount_amount is >0
df['coupon_used']= df['discount_amount']>0
df['coupon_used']
```

```python
# 1. What percentage of orders use coupons?
coupon_precentage = df['coupon_used'].mean() *100
print(f"coupon_percentage: {coupon_precentage:.2f}% of orders used coupon")
 
```

```python

# 2. Do coupon users spend more than non-coupon users?
spending_comparision = (
    df.groupby('coupon_used')['order_amount']
    .agg(
        Total_orders = 'count',
        avg_order_amount = 'mean',
        Total_spending= 'sum'
    )
    .reset_index()
)
spending_comparision['customer_type'] = (
    spending_comparision['coupon_used']
    .map(
        {True : 'coupon users',
        False : "non_coupon_users"})
)
spending_comparision
```

```python
# 3. Which city uses the most coupons?
coupon_by_city = (
    df[df['coupon_used']]
    .groupby('city_y')
    .size()
    .sort_values(ascending =False)
    .reset_index(name= 'coupon_orders')

)
coupon_by_city
```

```python

# 4. Are coupons helping customer retention?

df['order_timestamp'] = pd.to_datetime(df['order_timestamp'])

customer_orders = df.sort_values(
    ['customer_id','order_timestamp']
)
customer_retention = (
    customer_orders.groupby('customer_id')
    .agg(
        total_orders = ('order_id' , 'count'),
        first_order_coupon = ('coupon_used','first')
    )
    .reset_index()

)
customer_retention['retained'] = (
    customer_retention['total_orders']>1

)
retention_comparision = (
    customer_retention.groupby('first_order_coupon')
    .agg(
        total_customers= ('customer_id','count'),
        retained_customers = ('retained' , 'sum'),
        retention_rate = ('retained','mean')
    )
    .reset_index()
)

retention_comparision['retention_rate'] *=100
retention_comparision['customer_type'] =(
    retention_comparision ['first_order_coupon'].map({
        True : 'first_order coupon users',
        False: 'first_order non -coupon users'
    })
)

retention_comparision
```
